```python
from __future__ import annotations

import warnings

from ase.md.velocitydistribution import MaxwellBoltzmannDistribution
from pymatgen.core import Lattice, Structure
from pymatgen.io.ase import AseAtomsAdaptor
from ase.io import read
import pandas as pd

import matgl
matgl.set_backend("DGL")
from matgl.ext._ase_dgl import MolecularDynamics, PESCalculator, Relaxer

# To suppress warnings for clearer output
warnings.simplefilter("ignore")

pot = matgl.load_model("../pretrained_models/QET-MatQ-PES")
```

```python
atoms = read("./solvation_zn/zn_300_100.000ps.xyz")

calc = PESCalculator(pot)
# set up the calculator for atoms object
atoms.set_calculator(calc)
df = pd.DataFrame({"symbols": atoms.get_chemical_symbols(), "charges": atoms.get_charges()})
df.groupby(by=["symbols"]).mean()
```
